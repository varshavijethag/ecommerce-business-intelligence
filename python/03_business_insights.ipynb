```python
import pandas as pd

monthly_sales = pd.read_csv("../data/cleaned/monthly_sales.csv")
yearly_sales = pd.read_csv("../data/cleaned/yearly_sales.csv")
category_sales = pd.read_csv("../data/cleaned/category_sales.csv")

print("Sales data loaded successfully!")
```

```python
best_month = monthly_sales.loc[
    monthly_sales["sales"].idxmax()
]

best_month
```

```python
worst_month = monthly_sales.loc[
    monthly_sales["sales"].idxmin()
]

worst_month
```

```python
sales_2017 = yearly_sales.loc[
    yearly_sales["year"] == 2017, "sales"
].iloc[0]

sales_2018 = yearly_sales.loc[
    yearly_sales["year"] == 2018, "sales"
].iloc[0]

growth_2018 = ((sales_2018 - sales_2017) / sales_2017) * 100

print("Sales Growth from 2017 to 2018:", round(growth_2018, 2), "%")
```

```python
best_category = category_sales.loc[
    category_sales["sales"].idxmax()
]

best_category
```

```python
product_sales = pd.read_csv("../data/cleaned/product_sales.csv")

best_product = product_sales.loc[
    product_sales["sales"].idxmax()
]

best_product
```

```python
payment_summary = pd.read_csv("../data/cleaned/payment_summary.csv")

best_payment = payment_summary.loc[
    payment_summary["order_count"].idxmax()
]

best_payment
```

```python
order_status = pd.read_csv("../data/cleaned/order_status_summary.csv")

order_status
```

```python
delivered_orders = order_status.loc[
    order_status["order_status"] == "delivered",
    "order_count"
].iloc[0]

total_status_orders = order_status["order_count"].sum()

delivered_rate = (delivered_orders / total_status_orders) * 100

print("Delivered Order Rate:", round(delivered_rate, 2), "%")
```
